```python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
```

BASE TASK 1

```python
cities_clean = [city.strip().capitalize() for city in cities]
print(*cities_clean)
print(cities_clean)
```

BASE TASK 2


```python
amounts_list = [int(a) for a in amounts]
print(*amounts_list)
print(amounts_list)
```

BASE TASK 3

```python
total = sum(amounts_list)
total_check = len(amounts_list)
avg_sum = round(total / total_check)

print(total)
print(total_check)
print(avg_sum)
```

BASE TASK 4

```python
cnt = 0
sum_notpaid = 0
for i in range(len(amounts_list)):
  if paid[i] == False:
    cnt += 1
    sum_notpaid += amounts_list[i]

print(sum_notpaid)
print(cnt)
```

BASE TASK 5

```python
print(f"{'Город':<12} | {'Сумма':>7} | {'Статус':<10}")
print("-" * 36)
for city, amount, is_paid in zip(cities_clean, amounts_list, paid):
  status = "Оплачено" if is_paid else "Не оплачено"

  print(f"{city:<12} | {amount:>7} | {status:<10}")
```

**STANDART PART**

STANDART TASK 6

```python
orders = [
    {
        'city': city,
        'amount': amount,
        'paid': is_paid,
        'day': day
    }
    for city, amount, is_paid, day in zip(cities_clean, amounts_list, paid, days)
]

total_almaty = 0
total_astana = 0
for i in orders:
  if i['city'] == 'Almaty' and i['paid'] == True:
    total_almaty += i['amount']
  elif i['city'] == 'Astana' and i['paid'] == True:
    total_astana += i['amount']
  else:
    pass

print(f"Выручка оплаченая по Алматы: {total_almaty}")
print(f"Выручка оплаченая по Астане: {total_astana}")
```

STANDART TASK 7

```python
max_id, max_paid = max(((idx, item) for idx, item in enumerate(orders) if item['paid']),
                       key=lambda x: x[1]['amount'])

result_index = max_id
result_city = max_paid['city']
print(result_index)
print(result_city)
```

STANDART TASK 8

```python
all_orders = len(days)

weekend_days = len([day for day in days if day == 'сб' or day == 'вс'])
percentage_wknd_days = round((weekend_days / all_orders) * 100)

print(percentage_wknd_days)
```

STANDART TASK 9

```python
income_month_1 = 15000
course_price = 400000

month_counter_1 = 1
while(course_price > income_month_1):
  month_counter_1 +=1
  income_month_1 += 15000

month_counter_2 = 1
saved_money = 15000
current_saving = 15000
while(course_price > saved_money):
  month_counter_2 += 1
  current_saving += 1000
  saved_money += current_saving

print(month_counter_1)
print(month_counter_2)
```
