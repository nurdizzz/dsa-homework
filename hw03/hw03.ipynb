```python
import csv
import numpy as np

CITY = {"алматы": "Алматы", "almaty": "Алматы", "астана": "Астана", "astana": "Астана",
        "шымкент": "Шымкент", "shymkent": "Шымкент", "караганда": "Караганда",
        "актобе": "Актобе", "атырау": "Атырау"}

with open("orders.csv", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))

print(rows[:5])

# функции из урока 2 + превращаем колонки в массивы NumPy
amount = np.array([float(r["amount"].replace(" ", "")) for r in rows])
paid   = np.array([r["status"] == "оплачен" for r in rows])
days   = np.array([float(r["delivery_days"]) if r["delivery_days"] else np.nan for r in rows])
city   = np.array([CITY.get(r["city"].strip().lower(), r["city"].strip()) for r in rows])
cust   = np.array([r["customer_id"] for r in rows])

print(amount.shape, amount.dtype, paid.dtype)
```

База TASK 1

```python
max_amount = np.argmax(amount)
max_sum = amount[max_amount]
max_sum_client = cust[max_amount]
max_sum_city = city[max_amount]

print(max_amount)
print(max_sum)
print(max_sum_client)
print(max_sum_city)
```

База TASK 2

```python
paid_sum = amount[paid].sum()
not_paid_sum = amount[~paid].sum()

print(paid_sum)
print(not_paid_sum)
```

База TASK 3

```python
shym_orders = np.array([a for a, cit in zip(amount, city) if cit == "Шымкент"])
all_orders = len(amount)
percentage_of_shym_orders = (len(shym_orders) / all_orders) * 100
print(len(shym_orders))
print(percentage_of_shym_orders)
```

Стандарт TASK 4

```python
large_order = amount >= 50000
large_and_paid = large_order & paid
without_test = large_and_paid & (cust != 'TEST')

total = amount[large_and_paid].sum() * 0.15
total_without_test = amount[without_test].sum() * 0.15
print(total)
print(total_without_test)
```

Стандарт TASK 5

```python
unique_cities = np.unique(city)

for c in unique_cities:
  city_mask = (city == c)
  fast_delivery_mask = city_mask & (days >= 1) & (days <= 2)

  total_city_orders = city_mask.sum()
  total_near_city_orders = fast_delivery_mask.sum()
  perc_of_orders = (total_near_city_orders / total_city_orders) * 100
  print(f'{c}: {perc_of_orders}')


```
