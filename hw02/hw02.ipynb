r = {}
for c in ["A", "B", "A"]:
    r[c] = r.get(c, 0) + 1
print(r)

Ответ будет [2, 1]
Обьяснение: Вначале инициируем пустой словарье(его еще заполним), после чего прогоняем форик по списку ["A", "B", "A"], внутри даем значение в r[c] (пока что пустой). r.get(c, 0) + 1 это у нас значение - если в словаре есть ключ под С то добавляем + 1, если же нет создаем и задаем дефолтное значение 0 и также прибавляем +1

```python
r = {}
for c in ["A", "B", "A"]:
    r[c] = r.get(c, 0) + 1
print(r)
```

```python
import csv
with open("orders.csv", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))
print(len(rows), "строк")
print(rows[1:5])
```

Task 1

```python
def to_nums(s):
  return s.strip().replace(' ', '')

def clean_city(s):
  return s.capitalize()
```

Task 2

```python
order_chart = {}

for row in rows:
  ctg = row["category"]
  order_chart[ctg] = order_chart.get(ctg, 0) + 1

sorted_chart = dict(sorted(order_chart.items(), key=lambda item: item[1], reverse=True))
print(sorted_chart)
```

Task 3

```python
stats = {}

for row in rows:
    if row['status'] == 'оплачен':
        cat = row['category']
        amount = float(to_nums(row['amount']))

        if cat not in stats:
            stats[cat] = {'total': 0.0, 'count': 0}

        stats[cat]['total'] += amount
        stats[cat]['count'] += 1

header = f"{'Категория':<12} | {'Выручка':<12} | {'Средний чек':<12}"
print(header)
print("-" * len(header))

for cat, data in stats.items():
    total_rev = data['total']
    avg_check = total_rev / data['count']
    print(f"{cat:<12} | {total_rev:<12.1f} | {avg_check:<12.1f}")
```

Task 4

```python
months_count = {}

for row in rows:
    month = row['order_date'][:7]
    months_count[month] = months_count.get(month, 0) + 1

for month, count in months_count.items():
    print(f"Месяц: {month} | Заказов: {count}")
```
